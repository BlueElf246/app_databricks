# Gold — shared UDFs

Keyword normalization used by `dim_keyword` and `fact_search_hit` (rule from design 1.3). Load it with `%run ./_udfs`.

- `normalize_keyword`: NFC, lower-case, trimmed, single spaces, **diacritics kept** ("phở" ≠ "pho").
- `unaccent_vi`: diacritics removed (đ → d), only for fuzzy matching.


```python
import re
import unicodedata

from pyspark.sql.types import StringType


def normalize_keyword(s):
    if s is None:
        return None
    return re.sub(r"\s+", " ", unicodedata.normalize("NFC", s).lower().strip())


def unaccent_vi(s):
    if s is None:
        return None
    s = s.replace("đ", "d").replace("Đ", "D")
    return "".join(ch for ch in unicodedata.normalize("NFD", s) if unicodedata.category(ch) != "Mn")


spark.udf.register("normalize_keyword", normalize_keyword, StringType())
spark.udf.register("unaccent_vi", unaccent_vi, StringType())
print("UDFs registered: normalize_keyword, unaccent_vi ✅")
```
